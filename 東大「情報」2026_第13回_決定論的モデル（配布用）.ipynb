#マルサスの人口モデル

```python
def generate_data_from_Malthus(T, m, x0):
  x = [0] * T
  x[0] = x0

  for t in range(1,T):
    x[t] = m * x[t-1]

  return x

generate_data_from_Malthus(10, 0.9954, 1.2*10**8)
```

#ロジスティックモデル

```python
import numpy as np
import matplotlib.pyplot as plt

m = 1.2      # 基本的な増大率
b = 0.001    # 人口が増えたときに成長率がどれくらい抑えられるか
x0 = 10      # 初期人口
T = 100      # シミュレーション期間

# 結果を保存するリスト
x = np.zeros(T + 1)
x[0] = x0

# シミュレーション
for t in range(1, T + 1):
    x[t] = m * (1 - b * x[t-1]) * x[t-1]

# 下記のコメントを外すと各年の人口が表示される（鬱陶しいので省略したが理解には役立つかも）
# for t in range(T + 1):
#     print(t, x[t])

plt.figure(figsize=(8, 5))
plt.plot(range(T + 1), x, marker="o")
plt.xlabel("Time")
plt.ylabel("Population")
plt.title("Malthus-type population model with decreasing growth rate")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate_population(m, b, x0, T):
    x = np.zeros(T + 1)
    x[0] = x0

    for t in range(1, T + 1):
        x[t] = m * (1 - b * x[t-1]) * x[t-1]

    return x

# 共通設定
x0 = 10
T = 100

# 複数のパラメータを試す
settings = [
    [1.1, 0.001],
    [1.2, 0.001],
    [1.5, 0.001],
    [1.2, 0.002],
]

plt.figure(figsize=(8, 5))

for m, b in settings:
    x = simulate_population(m, b, x0, T)
    plt.plot(range(T + 1), x, label=f"m={m}, b={b}")

plt.xlabel("Time")
plt.ylabel("Population")
plt.title("Comparison of population dynamics")
plt.legend()
plt.grid(True)
plt.show()
```
